```python
# Para reemplazar texto de archivos csv

import csv

archivo_original = "datos.csv"
archivo_nuevo = "datos_nuevos.csv"

with open(archivo_original, "r", encoding="utf-8") as entrada, \
     open(archivo_nuevo, "w", newline="", encoding="utf-8") as salida:
    
    lector = csv.reader(entrada)
    escritor = csv.writer(salida)
    
    for fila in lector:
        # Reemplaza 'texto_viejo' por 'texto_nuevo' en cada celda de la fila
        fila_modificada = [celda.replace("texto_viejo", "texto_nuevo") for celda in fila]
        escritor.writerow(fila_modificada)
```

```python
# Para reemplazar texto en archivos excel

import pandas as pd

# 1. Leer el archivo de Excel
df = pd.read_excel("datos.xlsx")

# 2. Reemplazar el texto en todo el documento
df = df.replace("texto_viejo", "texto_nuevo", regex=True)

# 3. Guardar en un archivo nuevo sin el índice numérico
df.to_excel("datos_nuevos.xlsx", index=False)
```
